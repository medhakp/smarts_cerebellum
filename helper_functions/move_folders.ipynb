# Move folders

Function to move folders in directory

```python
import os
import shutil
import pandas as pd
```

```python
def move_folder(source_folder, dest_dir):

    if not os.path.exists(source_folder):
        print(f'Folder not found')
        return None # exit function if folder not found
    
    shutil.move(source_folder, dest_dir)
```

```python
p_df = pd.read_csv('/cifs/diedrichsen/data/smarts_cerebellum/participants_anat.tsv', sep = '\t')
```

```python
# move old iso_norm_mniSymm folders into a common folder (for each subject-week) (to be deleted).
    # this is to help keep our trash organized.
for i in range(0, p_df.shape[0]):

    p_id = p_df['ID'].iloc[i]
    week = (p_df['Week'].iloc[i]).strip() # sometimes have extra white spaces
    p_centre = (str(p_df['Centre'].iloc[i])).strip()

    subj_id = f'{p_centre.strip()}_{p_id}'

    source_dir = f'/cifs/diedrichsen/data/smarts_cerebellum/anatomicals/{subj_id}/{week}/iso_norm_mniSymm'
    dest_dir = f'/cifs/diedrichsen/data/smarts_cerebellum/prev_iso_norm_mniSymm/{subj_id}/{week}'

    if os.path.isdir(source_dir):
        move_folder(source_folder = source_dir, dest_dir = dest_dir)
        print(f'Moved folder for {subj_id} {week}')
```

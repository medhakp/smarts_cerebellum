```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import os
import numpy as np
from scipy import stats
import smarts_cerebellum.globals as gl

mcp = pd.read_csv(os.path.join(gl.baseDir, 'MNISymC_FaMap', 'summary_MNISymC_MCP_FaMap.tsv'), sep = '\t')
mcp = mcp[mcp.regionname == 'MCP_L']

cst = pd.read_csv(os.path.join(gl.baseDir, 'MNISymC_FaMap', 'summary_MNISymC_CST_FaMap.tsv'), sep = '\t')
cst = cst[cst.regionname == 'CST_R']
```

```python
fig, axs = plt.subplots(1, 5, figsize = (9,2), sharey = True, sharex = True, constrained_layout = True)
for i, week in enumerate(gl.weeks):
    mcp_week = mcp[mcp.Week == week]
    cst_week = cst[cst.Week == week]

    corr = pd.merge(mcp_week, cst_week, on = ['subj_id', 'Week'],suffixes = ('_mcp', '_cst'))

    ax = axs[i]
    sns.regplot(x = 'asymmetry_index_cst', y = 'asymmetry_index_mcp', data = corr, ax = ax, color = 'hotpink')

    ax.set_ylabel('')
    ax.set_xlabel('')

    ax.set_title(f'week {week}')

axs[0].set_ylabel("mcp")
axs[0].set_xlabel(f"cst")
fig.suptitle("correlations btwn cst and mcp (FA asymmetry index)")

sns.despine(trim = True)
plt.show()
```

```python
print("---pearson R correlation--- \n")
for week in gl.weeks:
    mcp_week = mcp[mcp.Week == week]
    cst_week = cst[cst.Week == week]
    corr = pd.merge(mcp_week, cst_week, on = ['subj_id', 'Week'],suffixes = ('_mcp', '_cst'))
    r_stat, p_stat = stats.pearsonr(corr['asymmetry_index_cst'], corr['asymmetry_index_mcp'])
    print(f'week {week}: r = {r_stat}, p-value = {p_stat}')
```

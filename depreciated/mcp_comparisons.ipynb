```python
import pandas as pd
import os
import smarts_cerebellum.globals as gl
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import numpy as np

# mcp_ventral_pons metrics
mcp_ventral_pons = pd.read_csv(os.path.join(gl.baseDir, 'MNISymC_WM', 'summary_MNISymC_MCP_with_ventral_pons_WM_mod.tsv'), sep = '\t')
#mcp_ventral_pons = mcp_ventral_pons[mcp_ventral_pons.regionname == 'CST_R']
mcp_ventral_pons = mcp_ventral_pons[mcp_ventral_pons.isPatient == 1]

# mcp metrics
mcp = pd.read_csv(os.path.join(gl.baseDir, 'MNISymC_WM', 'summary_MNISymC_MCP_WM_mod.tsv'), sep = '\t')
#mcp = mcp[mcp.regionname == 'CST_R']
mcp = mcp[mcp.isPatient == 1]
```

```python
fig, axs = plt.subplots(1, 2, figsize = (6, 3), sharey = True, sharex = True, constrained_layout = True)
sns.pointplot(x = 'Week', y = 'mean', data = mcp_ventral_pons[mcp_ventral_pons.regionname == 'MCP_with_ventral_pons_R'], color = 'lightblue', label = 'ipsi with pons', ax = axs[0], errorbar = 'se', native_scale = True)
sns.pointplot(x = 'Week', y = 'mean', data = mcp[mcp.regionname == 'MCP_R'], color = 'blue', label = 'ipsi no pons', ax = axs[0], native_scale = True, errorbar = 'se')

sns.pointplot(x = 'Week', y = 'mean', data = mcp_ventral_pons[mcp_ventral_pons.regionname == 'MCP_with_ventral_pons_L'], color = 'salmon', label = 'contra with pons', ax = axs[1], errorbar = 'se', native_scale = True)
sns.pointplot(x = 'Week', y = 'mean', data = mcp[mcp.regionname == 'MCP_L'], color = 'red', label = 'contra no pons', ax = axs[1], native_scale = True, errorbar = 'se')

for i in range(len(axs)):
    axs[i].set_xticks(gl.weeks)

fig.suptitle("WMV in MCP")
sns.despine(trim = True)
plt.show()
```

# Surface projection

of mean and median slope images (for each of patients and controls) from pipelines: MNISym_coreg_normalize and MNISym_coreg_regresssion

Authors: Marco,

```python
import SUITPy.flatmap as flatmap
import matplotlib.pyplot as plt
import smarts_cerebellum.globals as gl
import os
```

```python
space = 'MNISym'
stats = 'nanmean' # for discrete data; module (by-default) assumes continuous functional data

# in previous, we did vol_to_surf without stats = 'mode'

mean_path = os.path.join(gl.baseDir, 'Regression', 'mean_images')
median_path = os.path.join(gl.baseDir, 'Regression', 'median_images')
```

## Surface projections of GM

```python
datatype = 'GM'
```

```python
group = 'patients'

filename = f'{group}_{space}_{datatype}_coreg_slope_mean.nii'
funcdata = flatmap.vol_to_surf(mean_path + '/'+ filename,space=space, stats = stats)
flatmap.plot(data=funcdata, cmap='bwr', new_figure=True, colorbar=True, cscale=(-.005, .005))
plt.title(f'{group} {datatype} mean slope')
```

```python
group = 'controls'

filename = f'{group}_{space}_{datatype}_coreg_slope_mean.nii'
funcdata = flatmap.vol_to_surf(mean_path + '/'+ filename,space=space, stats = stats)
flatmap.plot(data=funcdata, cmap='bwr', new_figure=True, colorbar=True, cscale=(-.005, .005))
plt.title(f'{group} {datatype} mean slope')
```

It is possible that the controls' image looks like this because of the outlier we identified in analysis (CUP_1001).

Should perform that analysis and determine if there are outliers we should consider for our surface projections.

```python
group = 'patients'

filename = f'{group}_{space}_{datatype}_coreg_slope_median.nii'
funcdata = flatmap.vol_to_surf(median_path + '/'+ filename,space=space, stats = stats)
flatmap.plot(data=funcdata, cmap='bwr', new_figure=True, colorbar=True, cscale=(-.005, .005))
plt.title(f'{group} {datatype} median slope')
```

```python
group = 'controls'

filename = f'{group}_{space}_{datatype}_coreg_slope_median.nii'
funcdata = flatmap.vol_to_surf(median_path + '/'+ filename,space=space, stats = stats)
flatmap.plot(data=funcdata, cmap='bwr', new_figure=True, colorbar=True, cscale=(-.005, .005))
plt.title(f'{group} {datatype} median slope')
```

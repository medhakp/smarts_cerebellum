```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import smarts_cerebellum.globals as gl
```

```python
path = os.path.join(gl.baseDir, 'Regression')
df = pd.read_csv(f'{path}/MNISym_coreg_slope_rightLesion_NettekovenAsym32_summarized.tsv', sep = '\t')
df=df[df.subj_id != 'CUP_1001'] # outlier -> excluded
```

```python
df_GM = df[df.image == 2]
df_T1 = df[df.image == 1]
```

```python
sns.pointplot(data = df_GM, x='regionname', y='mean', hue = 'isPatient', errorbar = 'se', estimator = 'mean')

sns.despine(trim = True)
plt.title('Mean of mean GM slope in ROIs')
plt.show()
```

```python
sns.pointplot(data = df_T1, x='regionname', y='mean', hue = 'isPatient', errorbar = 'se', estimator = 'mean', alpha = 0.7, dodge = 0.1)

sns.despine(trim = True)
plt.title('Mean of mean T1 slope in ROIs')
plt.show()
```

```python
df.regionname.unique()
```

```python
# get motor ROIs
rois = []
for roi in df.regionname.unique():
    if roi[0]=='M':
        rois.append(roi)
```

```python
df_rois_GM = df[df.regionname.isin(rois) & (df.image == 2)]
df_rois_T1 = df[df.regionname.isin(rois) & (df.image == 1)]
df_rois_CSF = df[df.regionname.isin(rois) & (df.image == 4)]
```

```python
sns.boxplot(data=df_rois_GM, x='regionname', y='mean', hue='isPatient', showfliers = False)

sns.despine(trim = True)
plt.title('Mean of the mean GM slope in motor regions')
plt.show()
```

```python
sns.pointplot(data=df_rois_GM, x='regionname', y='mean', hue='isPatient', errorbar = 'se', estimator = 'mean')

sns.despine(trim = True)
plt.title('Mean of the mean GM slope in motor regions')
plt.show()
```

```python
sns.pointplot(data=df_rois_T1, x='regionname', y='mean', hue='isPatient', errorbar='se', estimator='mean', alpha = 0.7)

sns.despine(trim = True)
plt.title('Mean of the mean T1 slope in motor regions')
plt.show()
```

```python
sns.pointplot(data=df_rois_CSF, x='regionname', y='mean', hue='isPatient', errorbar='se', estimator='mean', alpha = 0.7)

sns.despine(trim = True)
plt.title('Mean of the mean CSF slope in motor regions')
plt.show()
```

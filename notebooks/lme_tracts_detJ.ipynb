```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
import smarts_cerebellum.globals as gl

obs = pd.read_csv(os.path.join(gl.baseDir, 'MNISymC_trans', 'summary_MNISymC_CST_T1_to-MNI152NLin2009cSymC_mode-image_detJ.tsv'), sep = '\t')
obs = obs[obs.isPatient == 1]

colours = ['r', 'blue']

fig, ax = plt.subplots(figsize = (3, 3), sharey = True, constrained_layout = True)
for i, H in enumerate(gl.Hem):
    obs_side = obs[obs.hemisphere.str.lower() == H]
    sns.pointplot(x = 'Week', y = 'mean', data = obs_side, errorbar = 'se', color = colours[i], native_scale = True, ax = ax, lw = 2,
                  label = 'ipsilesional' if H == 'r' else 'contralesional')



ax.set_xticks(gl.weeks)
ax.set_xlabel("Week")

# remove ticks for all plots except first on y-axis
ax.tick_params(axis = 'y', length = 0)

ax.set_ylabel('det(J)')

fig.suptitle('size of CST over 1 year')
sns.despine(trim = True)

plt.show()
```

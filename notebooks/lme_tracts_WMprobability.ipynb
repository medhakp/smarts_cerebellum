```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
import smarts_cerebellum.globals as gl

obs = pd.read_csv(os.path.join(gl.baseDir, 'MNISymC_WM', 'summary_MNISymC_CST_WM.tsv'), sep = '\t')
obs = obs[obs.isPatient == 1]

colours = ['r', 'blue']

fig, ax = plt.subplots(figsize = (3, 3), sharey = True, constrained_layout = True)
sns.pointplot(x = 'Week', y = 'mean', data = obs, hue = 'side', palette = {'ipsilesional': 'blue', 'contralesional': 'red'},
                errorbar = 'se', lw = 2, ax = ax, dodge = .3, native_scale = True )


ax.set_xticks(gl.weeks)
ax.set_xlabel("Week")
ax.set_ylim(.86, .98)

# remove ticks for all plots except first on y-axis
ax.tick_params(axis = 'y', length = 0)

ax.set_ylabel('probability')

fig.suptitle('P(WM) in CST over 1 year')
sns.despine(trim = True)

plt.show()
```

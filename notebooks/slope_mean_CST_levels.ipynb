```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
import pingouin as pg
import numpy as np
from scipy import stats
import os
import smarts_cerebellum.globals as gl
```

```python
midbrain_wm  = pd.read_csv(os.path.join(gl.baseDir, 'regression', 'summary_MNISymC_CST_midbrain_WM_mod_slope.tsv'), sep = '\t') 
midbrain_wm = midbrain_wm.groupby(['subj_id', 'regionname', 'group']).mean(numeric_only=True).reset_index()

pons_wm  = pd.read_csv(os.path.join(gl.baseDir, 'regression', 'summary_MNISymC_CST_pons_wm_mod_slope.tsv'), sep = '\t') 
pons_wm = pons_wm.groupby(['subj_id', 'regionname', 'group']).mean(numeric_only=True).reset_index()

medulla_wm  = pd.read_csv(os.path.join(gl.baseDir, 'regression', 'summary_MNISymC_CST_medulla_wm_mod_slope.tsv'), sep = '\t') 
medulla_wm = medulla_wm.groupby(['subj_id', 'regionname', 'group']).mean(numeric_only=True).reset_index()
```

```python
colours = ['green', 'r', 'blue']

fig, axs = plt.subplots(1, 3, figsize = (5, 2), sharex = True, sharey = True, constrained_layout = False)
sns.boxplot(data = midbrain_wm, y = 'mean', hue = 'group', showfliers = False, palette = colours, ax = axs[0], legend = False)
axs[0].set_title('midbrain')

sns.boxplot(data = pons_wm, y = 'mean', hue = 'group', showfliers = False, palette = colours, ax = axs[1], legend = False)
axs[1].set_title('pons')

sns.boxplot(data = medulla_wm, y = 'mean', hue = 'group', showfliers = False, palette = colours, ax = axs[2])
axs[2].set_title('medulla')

for ax in axs:
    ax.axhline(0, ls='--', lw=.8, color='k')

axs[-1].legend(bbox_to_anchor=(1, .5), loc='center left')
ax.set_xticks([])

axs[-1].set_ylabel('slope')

sns.despine(trim = True, bottom=True)
fig.suptitle('White matter volume in CST midbrain', y = 1.13)


plt.show()
```

```python
t_val, p_val = stats.ttest_ind(midbrain_wm[midbrain_wm.group=='ipsilesional']['mean'], midbrain_wm[midbrain_wm.group=='controls']['mean'])
print(f'**t-test for WM CST midbrain: lesion side in patients, controls** \n t-value: {t_val} \t p-value: {p_val}')
print('\n')

t_val, p_val = stats.ttest_ind(midbrain_wm[midbrain_wm.group=='contralesional']['mean'], midbrain_wm[midbrain_wm.group=='controls']['mean'])
print(f'**t-test for WM CST midbrain: non-lesion side in patients, controls** \n t-value: {t_val} \t p-value: {p_val}')
print('\n')

t_val, p_val = stats.ttest_rel(midbrain_wm[midbrain_wm.group=='contralesional']['mean'], midbrain_wm[midbrain_wm.group=='ipsilesional']['mean'])
print(f'**t-test for WM CST midbrain: within patients, left, right hemispheres** \n t-value: {t_val} \t p-value: {p_val}')
print('\n')
```

```python
t_val, p_val = stats.ttest_ind(pons_wm[pons_wm.group=='ipsilesional']['mean'], pons_wm[pons_wm.group=='controls']['mean'])
print(f'**t-test for WM CST pons: lesion side in patients, controls** \n t-value: {t_val} \t p-value: {p_val}')
print('\n')

t_val, p_val = stats.ttest_ind(pons_wm[pons_wm.group=='contralesional']['mean'], pons_wm[pons_wm.group=='controls']['mean'])
print(f'**t-test for WM CST pons: non-lesion side in patients, controls** \n t-value: {t_val} \t p-value: {p_val}')
print('\n')

t_val, p_val = stats.ttest_rel(pons_wm[pons_wm.group=='contralesional']['mean'], pons_wm[pons_wm.group=='ipsilesional']['mean'])
print(f'**t-test for WM CST pons: within patients, left, right hemispheres** \n t-value: {t_val} \t p-value: {p_val}')
print('\n')
```

```python
t_val, p_val = stats.ttest_ind(medulla_wm[medulla_wm.group=='ipsilesional']['mean'], medulla_wm[medulla_wm.group=='controls']['mean'])
print(f'**t-test for WM CST medulla: lesion side in patients, controls** \n t-value: {t_val} \t p-value: {p_val}')
print('\n')

t_val, p_val = stats.ttest_ind(medulla_wm[medulla_wm.group=='contralesional']['mean'], medulla_wm[medulla_wm.group=='controls']['mean'])
print(f'**t-test for WM CST medulla: non-lesion side in patients, controls** \n t-value: {t_val} \t p-value: {p_val}')
print('\n')

t_val, p_val = stats.ttest_rel(medulla_wm[medulla_wm.group=='contralesional']['mean'], medulla_wm[medulla_wm.group=='ipsilesional']['mean'])
print(f'**t-test for WM CST medulla: within patients, left, right hemispheres** \n t-value: {t_val} \t p-value: {p_val}')
print('\n')
```

```python
import pandas as pd
import os
import matplotlib.pyplot as plt
import seaborn as sns
import smarts_cerebellum.globals as gl

# currently avaialble: 'WM_mod', 'T1'
segment = 'WM'

# pons_cst means
pons_cst = pd.read_csv(os.path.join(gl.baseDir, f'MNISymC_{segment}', f'summary_MNISymC_CST_pons_{segment}_mod.tsv'), sep = '\t')
pons_cst = pons_cst[pons_cst.isPatient == 1]

# medulla_cst means
medulla_cst = pd.read_csv(os.path.join(gl.baseDir, f'MNISymC_{segment}', f'summary_MNISymC_CST_medulla_{segment}_mod.tsv'), sep = '\t')
medulla_cst = medulla_cst[medulla_cst.isPatient == 1]
```

```python
colours = ['red', 'blue']
fig, ax = plt.subplots(figsize=(3,2), sharey = True, constrained_layout=True)
sns.pointplot(x = 'Week', y = 'mean', data = pons_cst, hue = 'side', errorbar = 'se', palette = colours, native_scale = True)

sns.despine(trim = True)
ax.set_ylabel('volume (mm^3)')
fig.suptitle('WMV in CST at level of pons', y = 1.15)
plt.xticks(gl.weeks)
plt.show()
```

```python
colours = ['red', 'blue']
fig, ax = plt.subplots(figsize=(3,2), sharey = True, constrained_layout=True)
sns.pointplot(x = 'Week', y = 'mean', data = medulla_cst, hue = 'side', errorbar = 'se', palette = colours, native_scale = True)

sns.despine(trim = True)
ax.set_ylabel('volume (mm^3)')
plt.xticks(gl.weeks)
fig.suptitle('WMV in CST at level of medulla', y = 1.15)
plt.show()
```

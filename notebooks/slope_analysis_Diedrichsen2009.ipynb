```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import os
import smarts_cerebellum.globals as gl
```

```python
path = os.path.join(gl.baseDir, 'Regression')
df = pd.read_csv(f'{path}/MNISym_coreg_slope_rightLesion_AtlasSUIT_summarized.tsv', sep = '\t')
df=df[df.subj_id != 'CUP_1001'] # outlier -> excluded

rois = ['Left_I_IV', 'Right_I_IV', 'Left_V', 'Right_V', 'Left_VI', 'Right_VI']

df_rois_CSF = df[df.regionname.isin(rois) & (df.image==4)]
df_rois_GM = df[df.regionname.isin(rois) & (df.image==2)]
df_rois_WM = df[df.regionname.isin(rois) & (df.image==3)]
df_rois_T1 = df[df.regionname.isin(rois) & (df.image==1)]
```

```python
sns.pointplot(data=df_rois_GM, x='regionname', y='mean', hue='isPatient', errorbar='se', estimator='mean')

sns.despine(trim = True)
plt.title('Mean of the mean GM slope in regions')
plt.show()
```

```python
sns.pointplot(data=df_rois_WM, x='regionname', y='mean', hue='isPatient', errorbar='se', estimator='mean')

sns.despine(trim = True)
plt.title('Mean of the mean WM slope in regions')
plt.show()
```

```python
sns.pointplot(data=df_rois_CSF, x='regionname', y='mean', hue='isPatient', errorbar='se', estimator='mean')

sns.despine(trim = True)
plt.title('Mean of the mean CSF slope in regions')
plt.show()
```

```python
sns.pointplot(data=df_rois_T1, x='regionname', y='mean', hue='isPatient', errorbar='se', estimator='mean', alpha = 0.7)

sns.despine(trim = True)
plt.title('Mean of the mean T1 slope in regions')
plt.show()
```

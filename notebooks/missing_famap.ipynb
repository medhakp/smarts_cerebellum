```python
import pandas as pd
import os
import smarts_cerebellum.globals as gl
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import numpy as np

# DTI metrics
dti = pd.read_csv(os.path.join(gl.baseDir, 'MNISymC_FaMap', 'summary_MNISymC_CST_FaMap.tsv'), sep = '\t')
dti = dti[dti.regionname == 'CST_R']
dti = dti[dti.isPatient == 1]

# cerebellum GMV
cerebel = pd.read_csv(os.path.join(gl.baseDir, 'MNISymC_GM', 'summary_MNISymC_Nettekoven_2024_GM_mod.tsv'), sep = '\t')
cerebel = cerebel[cerebel.isPatient==1] # just patients
```

```python
for week in gl.weeks:
    fa_subjs = set(dti[dti.Week == week].subj_id.unique())
    cer_subjs = set(cerebel[cerebel.Week == week].subj_id.unique())
    print(f'week {week}: \n', cer_subjs - fa_subjs)
    print('number of missing subjs:', len(cer_subjs - fa_subjs))
    print('number of subjs already have in dti:', len(fa_subjs))
    print('\n')
```

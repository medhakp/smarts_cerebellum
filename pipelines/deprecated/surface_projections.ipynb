# Surface projection

of mean and median slope images (for each of patients and controls) from pipelines: MNISym_coreg_normalize and MNISym_coreg_regresssion

@Authors: Marco, 

```python
import SUITPy.flatmap as flatmap
import matplotlib.pyplot as plt
import nibabel as nb
import numpy as np
import nitools as nt
import os
import smarts_cerebellum.globals as gl
```

```python
space = 'MNISym'
datatype = 'WM'
stattype = 'mean'

mean_path = os.path.join(gl.baseDir, 'Regression', 'mean_images')
median_path = os.path.join(gl.baseDir, 'Regression', 'median_images')
```

```python
group = 'patients'

filename = f'{group}_{space}_{datatype}_coreg_slope_{stattype}.nii'
funcdata = flatmap.vol_to_surf(mean_path + '/'+ filename,space=space)
flatmap.plot(data=funcdata, cmap='bwr', new_figure=True, colorbar=True, cscale=(-.005, .005))
```

```python
group = 'patients'
stattype = 'median'
filename = f'{group}_{space}_{datatype}_coreg_slope_{stattype}.nii'
funcdata = flatmap.vol_to_surf(median_path + '/'+ filename,space=space)
flatmap.plot(data=funcdata, cmap='bwr', new_figure=True, colorbar=True, cscale=(-.005, .005))
plt.title("title")
```

```python
group = 'controls'
stattype = 'median'
filename = f'{group}_{space}_{datatype}_coreg_slope_{stattype}.nii'
funcdata = flatmap.vol_to_surf(median_path + '/'+ filename,space=space)
flatmap.plot(data=funcdata, cmap='bwr', new_figure=True, colorbar=True, cscale=(-.005, .005))
```

```python
group = 'patients'
datatype = 'logJac'

filename = f'{group}_{space}_{datatype}_coreg_slope_{stattype}.nii'
funcdata = flatmap.vol_to_surf(mean_path + '/'+ filename,space=space)
flatmap.plot(data=funcdata, cmap='bwr', new_figure=True, colorbar=True, cscale=(-.005, .005))
```

```python
group = 'controls'
datatype = 'logJac'

filename = f'{group}_{space}_{datatype}_coreg_slope_{stattype}.nii'
funcdata = flatmap.vol_to_surf(mean_path + '/'+ filename,space=space)
flatmap.plot(data=funcdata, cmap='bwr', new_figure=True, colorbar=True, cscale=(-.005, .005))
```

Regression on white matter segmentations in native space using (full-image) coregistered images.

```python
# regression on WM segmentations in native space
type = 'WM'

for subj in p_df['subj_id'].unique():

    # find each subject's reference image, and run it through the regression
    refT1 = (p_df.loc[(p_df['subj_id']==subj), 'RefT1'].iloc[0]).strip()
    ref_img = f'{base_dir}/anatomicals/{subj}/{refT1}/c2{subj}_{refT1}_T1.nii'

    print(f"Regression on {subj} \n")

    intercept_img, slope_img = regression.perform_regression_week(subj_id = subj,
                         reference_img = ref_img
                         )
    

    # if images exist, save them
    if intercept_img is not None and slope_img is not None:
        results_path = f'{base_dir}/Regression/{subj}'
        results_path = Path(results_path)

        # if directory does not exist, make it.

        
        nib.save(intercept_img, f'{results_path}/{subj}_native_{type}_coreg_intercept.nii.gz')
        nib.save(slope_img, f'{results_path}/{subj}_native_{type}_coreg_slope.nii.gz')
```
